```python
from pathlib import Path
import geopandas as gpd
import matplotlib.pyplot as plt
import rioxarray as rxr
from shapely import box
from matplotlib.colors import ListedColormap, BoundaryNorm
from matplotlib.patches import Patch
from matplotlib_map_utils import north_arrow, scale_bar
```

```python
gpkg_path = Path('../data/outputs/brazil.gpkg')
output_path = Path('../data/outputs/study_area.png')

country = gpd.read_file(
    gpkg_path,
    layer='1'
)

states = gpd.read_file(
    gpkg_path,
    layer='2'
)

para = states[
    states['name']=='Pará'
]
```

```python
label = rxr.open_rasterio(
    'label_test.tif',
    masked=True
).squeeze('band').rio.reproject(para.crs)
```

```python
xmin, ymin, xmax, ymax = label.rio.bounds()
study_area = box(xmin, ymin, xmax, ymax)

study_area = gpd.GeoDataFrame(
    geometry=[study_area],
    crs=label.rio.crs
)
```

```python
fig, ax = plt.subplot_mosaic(
    [
        ['country', 'labels'],
        ['study_area', 'labels']
    ],
    figsize=(12, 8),
    width_ratios=[1, 1.5],
    constrained_layout=True
)

# Country locator: top-left
country.plot(
    ax=ax['country'],
    facecolor='none',
    edgecolor='black'
)

para.plot(
    ax=ax['country'],
    color='grey'
)

ax['country'].set(
    title='(a) Pará in Brazil',
    xlabel='Longitude (°)',
    ylabel='Latitude (°)'
)

# Study-area locator: bottom-left
para.plot(
    ax=ax['study_area'],
    facecolor='none',
    edgecolor='black'
)

study_area.plot(
    ax=ax['study_area'],
    facecolor='none',
    edgecolor='black'
)

ax['study_area'].set(
    title='(b) Study Area in Pará',
    xlabel='Longitude (°)',
    ylabel='Latitude (°)'
)

# Reference labels: entire right column
cmap = ListedColormap(['darkgreen', 'orange', 'lightgrey'])
norm = BoundaryNorm([-0.5, 0.5, 1.5, 2.5], cmap.N)

im = label.plot.imshow(
    ax=ax['labels'],
    cmap=cmap,
    norm=norm,
    interpolation='nearest',
    add_colorbar=False
)

ax['labels'].legend(
    handles=[
        Patch(facecolor='darkgreen', label='Non-deforestation'),
        Patch(facecolor='orange', label='Deforestation'),
        Patch(facecolor='lightgrey', label='Clouds')
    ],
    loc='upper center',
    bbox_to_anchor=(0.5, -0.08),
    ncols=3,
    frameon=False,
    title='Reference Labels'
)

ax['labels'].set(
    title='(c) Study Area Reference Labels',
    xlabel='Longitude (°)',
    ylabel='Latitude (°)'
)

north_arrow(
    ax=ax['labels'],
    location='upper right',
    scale=0.4,
    rotation={'degrees': 0},
    label={'color': 'white'}
)

scale_bar(
    ax=ax['labels'],
    location='lower left',
    style='boxes',
    bar={
        'projection': label.rio.crs,
        'unit': 'km'
    },
    text={'textcolor': 'white'}
)

plt.savefig(
    output_path,
    dpi=300,
    bbox_inches='tight'
)

plt.show()
```
