# Neighbourhood population density maps

Build three Folium maps for Enschede, Zwolle and Deventer from CBS neighbourhood boundaries (2026) and population counts (2025). Run the cells in order.

Install the dependencies in your notebook environment with `python -m pip install geopandas pandas openpyxl folium`. Put the following inputs in `data/inputs/` and launch the notebook from the project root or its `notebooks/` directory.

| Input | Used for |
| --- | --- |
| `WijkBuurtkaart_2026_v0.gpkg` | Neighbourhood polygons from the `buurten` layer |
| `kwb_2025.xlsx` | Resident counts from the first worksheet |

The join uses neighbourhood codes present in both vintages and checks that the municipality agrees. Each map uses its own colour scale.

```python
from pathlib import Path

import folium
import geopandas as gpd
import pandas as pd
from IPython.display import display

project_root = Path.cwd()
if project_root.name == "notebooks":
    project_root = project_root.parent

input_dir = project_root / "data" / "inputs"
map_dir = project_root / "web" / "maps"
municipalities = ("Enschede", "Zwolle", "Deventer")
```

## 1. Load boundaries and population counts

```python
neighbourhoods = gpd.read_file(
    input_dir / "WijkBuurtkaart_2026_v0.gpkg",
    layer="buurten",
    columns=["bu_code", "bu_naam", "gm_naam"],
)
neighbourhoods = neighbourhoods.loc[
    neighbourhoods["gm_naam"].isin(municipalities)
    & neighbourhoods.geometry.notna()
    & ~neighbourhoods.geometry.is_empty
].rename(columns={"bu_naam": "name", "gm_naam": "city"})

population = pd.read_excel(
    input_dir / "kwb_2025.xlsx",
    usecols=["gwb_code_10", "gm_naam", "a_inw"],
    dtype={"gwb_code_10": "string"},
).rename(columns={
    "gwb_code_10": "bu_code",
    "gm_naam": "city",
    "a_inw": "population",
})

neighbourhoods["bu_code"] = neighbourhoods["bu_code"].astype("string").str.strip()
population["bu_code"] = population["bu_code"].str.strip()
```

## 2. Join by neighbourhood code and calculate density

Calculate polygon areas in Dutch RD New (EPSG:28992) and divide the population by the unrounded area in square kilometres. Exclude missing or negative counts and zero-area polygons. Convert the joined geometries to WGS 84 for Folium.

```python
stats = neighbourhoods.merge(
    population.dropna(subset=["bu_code"]),
    on=["bu_code", "city"],
    how="inner",
    validate="one_to_one",
).to_crs(epsg=28992)

stats["population"] = pd.to_numeric(stats["population"], errors="coerce")
stats["area_sqkm"] = stats.geometry.area / 1_000_000
stats = stats.loc[(stats["population"] >= 0) & (stats["area_sqkm"] > 0)].copy()
stats["pop_density"] = (stats["population"] / stats["area_sqkm"]).round(2)
stats = stats.to_crs(epsg=4326)

missing_cities = set(municipalities) - set(stats["city"])
if missing_cities:
    raise ValueError(f"No mappable neighbourhoods for: {', '.join(sorted(missing_cities))}")
```

## 3. Build and export the three Folium maps

```python
map_dir.mkdir(parents=True, exist_ok=True)

for city in municipalities:
    city_data = stats.loc[
        stats["city"] == city, ["bu_code", "name", "pop_density", "geometry"]
    ].copy()

    web_map = folium.Map()
    choropleth = folium.Choropleth(
        geo_data=city_data,
        data=city_data,
        columns=["bu_code", "pop_density"],
        key_on="feature.properties.bu_code",
        fill_color="RdYlGn_r",
        legend_name="Population Density (Inhabitants / km²)",
    ).add_to(web_map)

    folium.GeoJson(
        city_data,
        zoom_on_click=True,
        style_function=lambda feature: {"fillOpacity": 0, "weight": 0},
        highlight_function=lambda feature: {
            "fillColor": "black",
            "fillOpacity": 0.2,
            "color": "black",
            "weight": 1,
        },
        tooltip=folium.GeoJsonTooltip(
            fields=["name", "pop_density"],
            aliases=["Name", "Population Density"],
            localize=True,
        ),
    ).add_to(web_map)

    web_map.fit_bounds(choropleth.geojson.get_bounds())
    output_path = map_dir / f"{city.lower()}.html"
    web_map.save(str(output_path))
    print(f"{city} — {len(city_data)} neighbourhoods — {output_path.relative_to(project_root)}")
    display(web_map)
```
